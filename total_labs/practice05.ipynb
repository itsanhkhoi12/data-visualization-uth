# Bản đồ với Marker (Maps with Markers) <a id="6"></a>

Hãy tải xuống và import dữ liệu về các vụ việc của sở cảnh sát bằng phương thức `read_csv()` của *pandas*.

Tải xuống bộ dữ liệu và đọc nó vào một dataframe của *pandas*:

```python
df_incidents = pd.read_csv('https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/Data%20Files/Police_Department_Incidents_-_Previous_Year__2016_.csv')

print('Dataset downloaded and read into a pandas dataframe!')
```

Hãy xem qua năm phần tử đầu tiên trong bộ dữ liệu của chúng ta.

```python
df_incidents.head()
```

Như vậy mỗi dòng gồm 13 đặc trưng (feature):
> 1. **IncidntNum**: Số vụ việc
> 2. **Category**: Loại tội phạm hoặc vụ việc
> 3. **Descript**: Mô tả về tội phạm hoặc vụ việc
> 4. **DayOfWeek**: Ngày trong tuần mà vụ việc xảy ra
> 5. **Date**: Ngày mà vụ việc xảy ra
> 6. **Time**: Thời điểm trong ngày mà vụ việc xảy ra
> 7. **PdDistrict**: Quận của sở cảnh sát

> 13. **PdId**: ID của sở cảnh sát

Hãy tìm xem có bao nhiêu mục (entry) trong bộ dữ liệu của chúng ta.

```python
df_incidents.shape
```

Như vậy dataframe gồm 150,500 vụ phạm tội xảy ra trong năm 2016. Để giảm chi phí tính toán, hãy chỉ làm việc với 100 vụ việc đầu tiên trong bộ dữ liệu này.

```python
# get the first 100 crimes in the df_incidents dataframe
limit = 100
df_incidents = df_incidents.iloc[0:limit, :]
```

Hãy xác nhận rằng dataframe của chúng ta giờ chỉ gồm 100 vụ phạm tội.

```python
df_incidents.shape
```

Bây giờ khi đã giảm bớt dữ liệu một chút, hãy trực quan hóa nơi các vụ phạm tội này xảy ra trong thành phố San Francisco. Chúng ta sẽ dùng kiểu mặc định, và khởi tạo mức zoom là 12.

```python
# San Francisco latitude and longitude values
latitude = 37.77
longitude = -122.42
```

```python
# create map and display it
sanfran_map = folium.Map(location=[latitude, longitude], zoom_start=12)

# display the map of San Francisco
sanfran_map
```

Bây giờ hãy chồng vị trí của các vụ phạm tội lên bản đồ. Cách làm điều đó trong **Folium** là tạo một *feature group* với các feature và style riêng rồi thêm nó vào `sanfran_map`.

```python
# instantiate a feature group for the incidents in the dataframe
incidents = folium.map.FeatureGroup()

# loop through the 100 crimes and add each to the incidents feature group
for lat, lng, in zip(df_incidents.Y, df_incidents.X):
    incidents.add_child(
        folium.features.CircleMarker(
            [lat, lng],
            radius=5, # define how big you want the circle markers to be
            color='yellow',
            fill=True,
            fill_color='blue',
            fill_opacity=0.6
        )
    )

# add incidents to map
sanfran_map.add_child(incidents)
```

Bạn cũng có thể thêm một số văn bản pop-up sẽ hiển thị khi bạn di chuột qua một marker. Hãy làm cho mỗi marker hiển thị loại tội phạm khi di chuột qua.

```python
# instantiate a feature group for the incidents in the dataframe
incidents = folium.map.FeatureGroup()

# loop through the 100 crimes and add each to the incidents feature group
for lat, lng, in zip(df_incidents.Y, df_incidents.X):
    incidents.add_child(
        folium.features.CircleMarker(
            [lat, lng],
            radius=5, # define how big you want the circle markers to be
            color='yellow',
            fill=True,
            fill_color='blue',
            fill_opacity=0.6
        )
    )

# add pop-up text to each marker on the map
latitudes = list(df_incidents.Y)
longitudes = list(df_incidents.X)
labels = list(df_incidents.Category)

for lat, lng, label in zip(latitudes, longitudes, labels):
    folium.Marker([lat, lng], popup=label).add_to(sanfran_map)    
    
# add incidents to map
sanfran_map.add_child(incidents)
```

Thật tuyệt phải không? Giờ đây bạn có thể biết loại tội phạm nào đã xảy ra tại mỗi marker.

Nếu bạn thấy bản đồ quá chật chội với tất cả các marker này, có hai cách khắc phục vấn đề. Giải pháp đơn giản hơn là loại bỏ các marker vị trí này và chỉ thêm văn bản vào chính các circle marker như sau:

```python
# create map and display it
sanfran_map = folium.Map(location=[latitude, longitude], zoom_start=12)

# loop through the 100 crimes and add each to the map
for lat, lng, label in zip(df_incidents.Y, df_incidents.X, df_incidents.Category):
    folium.features.CircleMarker(
        [lat, lng],
        radius=5, # define how big you want the circle markers to be
        color='yellow',
        fill=True,
        popup=label,
        fill_color='blue',
        fill_opacity=0.6
    ).add_to(sanfran_map)

# show map
sanfran_map
```

Cách khắc phục đúng đắn hơn là nhóm các marker thành các cụm (cluster) khác nhau. Mỗi cụm khi đó được biểu diễn bằng số vụ phạm tội trong mỗi khu phố. Các cụm này có thể được xem như những vùng riêng biệt của San Francisco mà bạn có thể phân tích riêng.

Để thực hiện điều này, chúng ta bắt đầu bằng việc khởi tạo một đối tượng *MarkerCluster* và thêm tất cả các điểm dữ liệu trong dataframe vào đối tượng này.

```python
from folium import plugins

# let's start again with a clean copy of the map of San Francisco
sanfran_map = folium.Map(location = [latitude, longitude], zoom_start = 12)

# instantiate a mark cluster object for the incidents in the dataframe
incidents = plugins.MarkerCluster().add_to(sanfran_map)

# loop through the dataframe and add each data point to the mark cluster
for lat, lng, label, in zip(df_incidents.Y, df_incidents.X, df_incidents.Category):
    folium.Marker(
        location=[lat, lng],
        icon=None,
        popup=label,
    ).add_to(incidents)

# display map
sanfran_map
```

Hãy lưu ý rằng khi bạn thu nhỏ hết cỡ, tất cả các marker được nhóm vào một cụm duy nhất, *global cluster*, gồm 100 marker hay vụ phạm tội, tức là tổng số vụ phạm tội trong dataframe của chúng ta. Khi bạn bắt đầu phóng to, *global cluster* sẽ bắt đầu tách thành các cụm nhỏ hơn. Phóng to hết cỡ sẽ cho ra các marker riêng lẻ.

Để tạo bản đồ `Choropleth`, chúng ta cần một tệp GeoJSON xác định các khu vực/ranh giới của bang, hạt hoặc quốc gia mà chúng ta quan tâm. Trong trường hợp của chúng ta, vì chúng ta muốn tạo bản đồ thế giới, chúng ta cần một GeoJSON xác định ranh giới của tất cả các quốc gia trên thế giới. Để thuận tiện cho bạn, chúng tôi sẽ cung cấp tệp này, vì vậy hãy tải xuống nó. Hãy đặt tên là **world_countries.json**.

```python
# download countries geojson file
! wget --quiet https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/Data%20Files/world_countries.json
    
print('GeoJSON file downloaded!')
```

Bây giờ khi đã có tệp GeoJSON, hãy tạo một bản đồ thế giới có tâm tại các giá trị *latitude* và *longitude* **[0, 0]**, với mức zoom ban đầu là 2.

```python
world_geo = r'world_countries.json' # geojson file

# create a plain world map
world_map = folium.Map(location=[0, 0], zoom_start=2)
```

Và bây giờ để tạo bản đồ `Choropleth`, chúng ta sẽ dùng phương thức *choropleth* với các tham số chính sau:

1. `geo_data`, là tệp GeoJSON.
2. `data`, là dataframe chứa dữ liệu.
3. `columns`, biểu thị các cột trong dataframe sẽ được dùng để tạo bản đồ `Choropleth`.
4. `key_on`, là khóa hoặc biến trong tệp GeoJSON chứa tên của biến mà chúng ta quan tâm. Để xác định điều đó, bạn cần mở tệp GeoJSON bằng bất kỳ trình soạn thảo văn bản nào và ghi lại tên của khóa hoặc biến chứa tên các quốc gia, vì các quốc gia là biến chúng ta quan tâm. Trong trường hợp này, **name** là khóa trong tệp GeoJSON chứa tên các quốc gia. Lưu ý rằng khóa này phân biệt chữ hoa chữ thường (case_sensitive), vì vậy bạn cần truyền chính xác như nó tồn tại trong tệp GeoJSON.

```python
# generate choropleth map using the total immigration of each country to Canada from 1980 to 2013
world_map.choropleth(
    geo_data=world_geo,
    data=df_can,
    columns=['Country', 'Total'],
    key_on='feature.properties.name',
    fill_color='YlOrRd', 
    fill_opacity=0.7, 
    line_opacity=0.2,
    legend_name='Immigration to Canada'
)

# display map
world_map
```

Theo chú giải của bản đồ `Choropleth`, màu của một quốc gia càng đậm và càng gần màu đỏ thì số người nhập cư từ quốc gia đó càng cao. Theo đó, lượng nhập cư cao nhất trong suốt 33 năm (từ 1980 đến 2013) đến từ Trung Quốc, Ấn Độ và Philippines, tiếp theo là Ba Lan, Pakistan và, thật thú vị, là Hoa Kỳ.

Hãy lưu ý chú giải (legend) đang hiển thị một ranh giới hoặc ngưỡng (threshold) âm. Hãy khắc phục điều đó bằng cách tự định nghĩa các ngưỡng và bắt đầu từ 0 thay vì -6,918!

```python
world_geo = r'world_countries.json'

# create a numpy array of length 6 and has linear spacing from the minimum total immigration to the maximum total immigration
threshold_scale = np.linspace(df_can['Total'].min(),
                              df_can['Total'].max(),
                              6, dtype=int)
threshold_scale = threshold_scale.tolist() # change the numpy array to a list
threshold_scale[-1] = threshold_scale[-1] + 1 # make sure that the last value of the list is greater than the maximum immigration

# let Folium determine the scale.
world_map = folium.Map(location=[0, 0], zoom_start=2)
world_map.choropleth(
    geo_data=world_geo,
    data=df_can,
    columns=['Country', 'Total'],
    key_on='feature.properties.name',
    threshold_scale=threshold_scale,
    fill_color='YlOrRd', 
    fill_opacity=0.7, 
    line_opacity=0.2,
    legend_name='Immigration to Canada',
    reset=True
)
world_map
```
