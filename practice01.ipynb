## *pandas* nâng cao cơ bản: Lập chỉ mục và chọn dữ liệu (Indexing and Selection - slicing)<a id="6"></a>

### Chọn cột (Select Column)
**Có hai cách để lọc theo tên cột:**

Cách 1: Nhanh và dễ, nhưng chỉ dùng được nếu tên cột KHÔNG chứa dấu cách hoặc ký tự đặc biệt.
```python
    df.column_name               # returns series
```

Cách 2: Chắc chắn hơn, và có thể lọc theo nhiều cột.

```python
    df['column']                  # returns series
```

```python 
    df[['column 1', 'column 2']]  # returns dataframe
```
---

Ví dụ: Hãy thử lọc theo danh sách các quốc gia ('Country').

````python
df_can.Country  # returns a series
````

Hãy thử lọc theo danh sách các quốc gia ('Country') và dữ liệu của các năm: 1980 - 1985.

````python
df_can[['Country', 1980, 1981, 1982, 1983, 1984, 1985]] # returns a dataframe
# notice that 'Country' is string, and the years are integers. 
# for the sake of consistency, we will convert all column names to string later on.
````

### Chọn dòng (Select Row)

Có 2 cách chính để chọn dòng:

```python
    df.loc[label]    # filters by the labels of the index/column
    df.iloc[index]   # filters by the positions of the index/column
```

Trước khi tiếp tục, hãy lưu ý rằng chỉ mục mặc định của bộ dữ liệu là một dãy số từ 0 đến 194. Điều này khiến việc truy vấn theo một quốc gia cụ thể rất khó khăn. Ví dụ, để tìm dữ liệu của Nhật Bản (Japan), chúng ta cần biết giá trị chỉ mục tương ứng.

Điều này có thể được khắc phục rất dễ dàng bằng cách đặt cột 'Country' làm chỉ mục (index) bằng phương thức `set_index()`.

````python
df_can.set_index('Country', inplace=True)
# tip: The opposite of set is reset. So to reset the index, we can use df_can.reset_index()
````

````python
df_can.head(3)
````

````python
# optional: to remove the name of the index
df_can.index.name = None
````

Ví dụ: Hãy xem số người nhập cư từ Nhật Bản (dòng 87) trong các trường hợp sau:
    1. Toàn bộ dữ liệu của dòng (tất cả các cột)
    2. Riêng năm 2013
    3. Các năm từ 1980 đến 1985

````python
# 1. the full row data (all columns)
df_can.loc['Japan']
````

````python
# alternate methods
df_can.iloc[87]
````

````python
df_can[df_can.index == 'Japan']
````

````python
# 2. for year 2013
df_can.loc['Japan', 2013]
````

````python
# alternate method
# year 2013 is the last column, with a positional index of 36
df_can.iloc[87, 36]
````

````python
# 3. for years 1980 to 1985
df_can.loc['Japan', [1980, 1981, 1982, 1983, 1984, 1984]]
````

````python
# Alternative Method
df_can.iloc[87, [3, 4, 5, 6, 7, 8]]
````

Các tên cột là số nguyên (chẳng hạn như các năm) có thể gây ra một số nhầm lẫn. Ví dụ, khi chúng ta tham chiếu đến năm 2013, có thể bị nhầm với chỉ mục vị trí thứ 2013.

Để tránh sự mơ hồ này, hãy chuyển các tên cột sang kiểu chuỗi (string): '1980' đến '2013'.

````python
df_can.columns = list(map(str, df_can.columns))
# [print (type(x)) for x in df_can.columns.values] #<-- uncomment to check type of column headers
````

````python
# useful for plotting later on
years = list(map(str, range(1980, 2014)))
years
````

### Lọc dữ liệu dựa trên một tiêu chí
Để lọc dataframe dựa trên một điều kiện, chúng ta chỉ cần truyền điều kiện đó dưới dạng một vector boolean. 

Ví dụ, hãy lọc dataframe để hiển thị dữ liệu của các quốc gia châu Á (AreaName = Asia).

````python
# 1. create the condition boolean series
condition = df_can['Continent'] == 'Asia'
print(condition)
````

````python
# 2. pass this condition into the dataFrame
df_can[condition]
````

````python
# we can pass multiple criteria in the same line.
# let's filter for AreaNAme = Asia and RegName = Southern Asia

df_can[(df_can['Continent']=='Asia') & (df_can['Region']=='Southern Asia')]

# note: When using 'and' and 'or' operators, pandas requires we use '&' and '|' instead of 'and' and 'or'
# don't forget to enclose the two conditions in parentheses
````

**Biểu đồ đường (line plot) là gì và tại sao nên dùng?**

Biểu đồ đường (line chart hay line plot) là một loại biểu đồ hiển thị thông tin dưới dạng một chuỗi các điểm dữ liệu gọi là 'marker' được nối với nhau bằng các đoạn thẳng. Đây là loại biểu đồ cơ bản, phổ biến trong nhiều lĩnh vực.
Hãy dùng biểu đồ đường khi bạn có một bộ dữ liệu liên tục. Loại biểu đồ này phù hợp nhất để trực quan hóa xu hướng của dữ liệu theo thời gian.

**Hãy bắt đầu với một nghiên cứu tình huống:**

Năm 2010, Haiti hứng chịu một trận động đất thảm khốc có cường độ 7.0. Trận động đất gây ra sự tàn phá và thiệt hại về người trên diện rộng, và khoảng ba triệu người bị ảnh hưởng bởi thảm họa thiên nhiên này. Là một phần trong nỗ lực nhân đạo của Canada, Chính phủ Canada đã tăng cường nỗ lực tiếp nhận người tị nạn từ Haiti. Chúng ta có thể nhanh chóng trực quan hóa nỗ lực này bằng biểu đồ `Line`:

**Câu hỏi:** Vẽ biểu đồ đường về lượng nhập cư từ Haiti bằng `df.plot()`.

Đầu tiên, chúng ta sẽ trích xuất chuỗi dữ liệu (data series) của Haiti.

````python
haiti = df_can.loc['Haiti', years] # passing in years 1980 - 2013 to exclude the 'total' column
haiti.head()
````

Tiếp theo, chúng ta sẽ vẽ biểu đồ đường bằng cách thêm `.plot()` vào dataframe `haiti`.

````python
haiti.plot()
````

*pandas* tự động điền các giá trị chỉ mục (các năm) vào trục x, và các giá trị cột (dân số) vào trục y. Tuy nhiên, hãy lưu ý rằng các năm không được hiển thị vì chúng có kiểu *string*. Do đó, hãy đổi kiểu của các giá trị chỉ mục sang *integer* để vẽ biểu đồ.

Ngoài ra, hãy gán nhãn cho trục x và trục y bằng `plt.title()`, `plt.ylabel()` và `plt.xlabel()` như sau:

````python
haiti.index = haiti.index.map(int) # let's change the index values of Haiti to type integer for plotting
haiti.plot(kind='line')

plt.title('Immigration from Haiti')
plt.ylabel('Number of immigrants')
plt.xlabel('Years')

plt.show() # need this line to show the updates made to the figure
````

Chúng ta có thể thấy rõ số người nhập cư từ Haiti tăng vọt từ năm 2010 khi Canada tăng cường nỗ lực tiếp nhận người tị nạn từ Haiti. Hãy chú thích (annotate) mức tăng vọt này trên biểu đồ bằng phương thức `plt.text()`.

````python
haiti.plot(kind='line')

plt.title('Immigration from Haiti')
plt.ylabel('Number of Immigrants')
plt.xlabel('Years')

# annotate the 2010 Earthquake. 
# syntax: plt.text(x, y, label)
plt.text(2000, 6000, '2010 Earthquake') # see note below

plt.show() 
````

Chỉ với vài dòng code, bạn đã có thể nhanh chóng nhận diện và trực quan hóa mức tăng vọt về nhập cư!

Lưu ý nhanh về các giá trị x và y trong `plt.text(x, y, label)`:
    
     Vì trục x (các năm) có kiểu 'integer', chúng ta chỉ định x là một năm. Trục y (số người nhập cư) có kiểu 'integer', nên chúng ta chỉ cần chỉ định giá trị y = 6000.
    
```python
    plt.text(2000, 6000, '2010 Earthquake') # years stored as type int
```
    Nếu các năm được lưu dưới dạng kiểu 'string', chúng ta cần chỉ định x là vị trí chỉ mục của năm đó. Ví dụ, chỉ mục thứ 20 là năm 2000 vì đó là năm thứ 20 tính từ năm cơ sở 1980.
```python
    plt.text(20, 6000, '2010 Earthquake') # years stored as type int
```
    Chúng ta sẽ tìm hiểu các phương pháp chú thích nâng cao trong các module sau.

Chúng ta có thể dễ dàng thêm nhiều quốc gia hơn vào biểu đồ đường để so sánh một cách có ý nghĩa về lượng nhập cư từ các quốc gia khác nhau. 

**Câu hỏi:** Hãy so sánh số người nhập cư từ Ấn Độ (India) và Trung Quốc (China) từ năm 1980 đến 2013.

Bước 1: Lấy bộ dữ liệu của Trung Quốc và Ấn Độ, rồi hiển thị dataframe.

````python
### type your answer here
df_CI = df_can.loc[['India','China'],years]
df_CI
````

<details><summary>Nhấn vào đây để xem gợi ý về các hàm cần dùng</summary>

### 💡 Gợi ý

Bạn có thể cân nhắc các hàm sau:

- <code style="color:red">loc[]</code> — chọn các dòng theo tên quốc gia (truyền một **danh sách** tên) và các cột theo biến `years`.

</details>

Bước 2: Vẽ biểu đồ. Chúng ta sẽ chỉ định rõ ràng là biểu đồ đường bằng cách truyền tham số `kind` vào `plot()`.

````python
### type your answer here
df_CI.plot(kind='line')
````

<details><summary>Nhấn vào đây để xem gợi ý về các hàm cần dùng</summary>

### 💡 Gợi ý

Bạn có thể cân nhắc các hàm sau:

- <code style="color:red">plot()</code> — gọi trên dataframe `df_CI` và chỉ rõ loại biểu đồ bằng tham số `kind`.

</details>

Trông chưa đúng lắm...

Hãy nhớ rằng *pandas* vẽ các chỉ mục (index) lên trục x và các cột (columns) thành những đường riêng biệt trên trục y. Vì `df_CI` là một dataframe có `country` làm chỉ mục và `years` làm các cột, trước tiên chúng ta phải chuyển vị (transpose) dataframe bằng phương thức `transpose()` để hoán đổi dòng và cột.

````python
df_CI = df_CI.transpose()
df_CI.head()
````

*pandas* sẽ tự động vẽ hai quốc gia trên cùng một biểu đồ. Hãy tiếp tục vẽ dataframe đã chuyển vị mới. Nhớ thêm tiêu đề cho biểu đồ và gán nhãn cho các trục.

````python
### type your answer here
df_CI.plot(kind='line')

plt.title('Immigration from China and India (1980 - 2013)')
plt.xlabel('Years')
plt.ylabel('Number of Immigrants')

plt.show()
````

<details><summary>Nhấn vào đây để xem gợi ý về các hàm cần dùng</summary>

### 💡 Gợi ý

Bạn có thể cân nhắc các hàm sau:

- <code style="color:red">index.map()</code> — đổi kiểu index (các năm) sang số nguyên; hãy nhớ dataframe cần được chuyển vị ở bước trước.
- <code style="color:red">plot()</code> — vẽ biểu đồ đường với `kind='line'`.
- <code style="color:red">plt.title()</code> — đặt tiêu đề cho biểu đồ.
- <code style="color:red">plt.xlabel() / plt.ylabel()</code> — gán nhãn cho trục x và trục y.
- <code style="color:red">plt.show()</code> — hiển thị biểu đồ.

</details>

<br>Từ biểu đồ trên, chúng ta có thể quan sát thấy rằng Trung Quốc và Ấn Độ có xu hướng nhập cư rất giống nhau qua các năm.

Biểu đồ đường là một công cụ tiện dụng để hiển thị nhiều biến phụ thuộc theo một biến độc lập. Tuy nhiên, khuyến nghị chỉ nên có không quá 5-10 đường trên một biểu đồ; nhiều hơn mức đó sẽ khiến biểu đồ trở nên khó diễn giải.

**Câu hỏi:** Hãy so sánh xu hướng của 5 quốc gia hàng đầu đóng góp nhiều nhất vào lượng nhập cư vào Canada.

````python
### type your answer here
df_top5 = df_can.sort_values(by = 'Total', axis = 0, ascending = False).head(5)[years]
df_top5 = df_top5.transpose()
df_top5.plot(kind='line')

plt.title('Top 5 countries immigrating to Canada (1980–2013)')
plt.ylabel('Number of Immigrants')
plt.xlabel('Years')

plt.show
````

<details><summary>Nhấn vào đây để xem gợi ý về các hàm cần dùng</summary>

### 💡 Gợi ý

Bạn có thể cân nhắc các hàm sau:

- <code style="color:red">sort_values()</code> — sắp xếp `df_can` theo cột `Total`; chú ý `ascending` và `inplace`.
- <code style="color:red">head()</code> — lấy 5 dòng đầu sau khi sắp xếp.
- <code style="color:red">transpose()</code> — hoán đổi dòng/cột sau khi chọn các cột năm.
- <code style="color:red">index.map()</code> — đổi index sang số nguyên.
- <code style="color:red">plot()</code> — dùng thêm tham số `figsize` để phóng to biểu đồ.
- <code style="color:red">plt.title()</code> — đặt tiêu đề cho biểu đồ.
- <code style="color:red">plt.xlabel() / plt.ylabel()</code> — gán nhãn cho trục x và trục y.
- <code style="color:red">plt.show()</code> — hiển thị biểu đồ.

> **Mẹo:** Nhớ rằng `pandas` lấy index làm trục x và mỗi cột là một đường.

</details>

### Các loại biểu đồ khác

Chúc mừng bạn! Bạn đã học được cách xử lý dữ liệu bằng python và tạo biểu đồ đường bằng Matplotlib. Ngoài biểu đồ đường mặc định, còn có nhiều kiểu biểu đồ khác, tất cả đều có thể được truy cập bằng cách truyền từ khóa `kind` vào `plot()`. Danh sách đầy đủ các biểu đồ có sẵn như sau:

* `bar` cho biểu đồ cột dọc (vertical bar plot)
* `barh` cho biểu đồ cột ngang (horizontal bar plot)
* `hist` cho biểu đồ histogram
* `box` cho biểu đồ boxplot
* `kde` hoặc `density` cho biểu đồ mật độ (density plot)
* `area` cho biểu đồ vùng (area plot)
* `pie` cho biểu đồ tròn (pie plot)
* `scatter` cho biểu đồ phân tán (scatter plot)
* `hexbin` cho biểu đồ hexbin
