### Quy mô chênh lệch ~75 lần

Xét riêng bảng bán hàng chính, quy mô dao động từ **2.043 dòng** (`dev_012`) đến **152.739 dòng** (`dev_008`) — chênh **~75 lần**. Đây là lý do bắt buộc dùng **thang đo logarit \(\log_{10}\)** khi so sánh tổng thể, và lý do mọi so sánh về quy mô tuyệt đối đều không công bằng.

```python
# Minh hoạ: đếm số file theo định dạng và kiểm tra tổng khớp 80
from collections import Counter

# Phân rã 80 file nghiệp vụ theo định dạng lưu trữ
file_inventory = {
    "xlsx": 69,
    "csv": 10,
    "sqlite": 1,
}

dem = Counter(file_inventory)
tong = sum(dem.values())

print("Số file theo định dạng:")
for dinh_dang, n in dem.most_common():
    print(f"  {dinh_dang:<10}{n:>4} file  ({n/tong:>6.1%})")
print(f"  {'Tổng':<10}{tong:>4} file")

assert tong == 80, "Tổng số file nghiệp vụ phải bằng 80"
print("\n-> Không tính README, calendar, rulebook hay questions.jsonl.")
```

## Schema thực tế và sự khác biệt về Data Grain

*Cùng là dữ liệu bán hàng, nhưng một dòng có thể mang ý nghĩa khác nhau*

## Ba lớp khung nhìn

| Lớp | Nội dung |
|---|---|
| **Physical Schema** | File, sheet, table, cột, kiểu dữ liệu, khoá |
| **Logical Schema** | Thực thể và mối quan hệ giữa các thực thể |
| **Data Grain** | Một dòng dữ liệu đại diện cho đơn hàng, dòng hàng hay sự kiện nào |

## Hai trường hợp đối chiếu

| | **Trường hợp A · Hoá đơn** | **Trường hợp B · Hoá đơn và dòng hàng** |
|---|---|---|
| Bảng ghi | Bản ghi cấp hoá đơn | Bảng hoá đơn **1:N** + Bảng dòng hàng |
| Một dòng đại diện cho | một hoá đơn | một dòng hàng A, dòng hàng B |
| Ví dụ đối chiếu | `dev_009` — chỉ có dữ liệu cấp hoá đơn, không có chi tiết mặt hàng | Hàng A, Hàng B thuộc cùng một đơn |
| Mô hình quan hệ | — | 1:N (hoá đơn → dòng hàng) |

## Bảng schema mẫu (template)

| Field | Data Type | Vai trò |
|---|---|---|
| Tên cột thực tế số 1 | Kiểu từ source | Khoá / định danh nếu xác minh được |
| Tên cột thực tế số 2 | Kiểu từ source | Thời gian / giá trị |
| Tên cột thực tế số 3 | Kiểu từ source | Thuộc tính nghiệp vụ |

> Bảng schema mẫu (template) để minh hoạ cấu trúc, **không phải** bảng trích xuất đã xác minh.

## Rủi ro đếm trùng

- Nếu cộng trực tiếp tiền hoá đơn trên bảng đã nối, doanh thu có thể bị **nhân lên**.

### Bằng chứng suy luận ngành nghề

Nhãn ngành được suy ra bằng **bằng chứng kép**: từ khóa trong tên hàng hóa / danh mục mặt hàng / tên cột **đối chiếu** với ngành công bố trong tài liệu mô tả.

- **14/14 doanh nghiệp** có từ khóa nhận diện ngành khớp **100%** với ngành công bố.
- Trường hợp **mâu thuẫn** được ghi nhận minh bạch thay vì áp đặt: `dev_001`, `dev_010`, `dev_013` có tên mẫu hàng hoá chỉ tới ngành khác với ngành mô tả trong tài liệu — dùng **mẫu hàng hóa thực tế** làm căn cứ, đồng thời đánh dấu bằng chứng là *mâu thuẫn* thay vì *đồng thuận*.
- Nhãn được chuẩn hoá qua một bản đồ ngành ngắn, tránh lỗi rút gọn tùy tiện (chữ "Nhà" dùng cho cả *Nhà thuốc* lẫn *Nhà hàng*).

> **Lưu ý:** Tập 14 doanh nghiệp phản ánh sự đa dạng của bán lẻ và dịch vụ vừa/nhỏ, **không** phải một mẫu đại diện thống kê cho toàn bộ thị trường Việt Nam.

```python
# Minh hoạ: vì sao nối bảng 1-N làm doanh thu bị nhân lên
# Một hoá đơn có 3 dòng hàng -> khi nối, hoá đơn lặp 3 lần.

don_hang = [
    {"order_id": "A1", "line_amount": 100_000},
    {"order_id": "A1", "line_amount":  50_000},
    {"order_id": "A1", "line_amount":  30_000},
]
chi_tiet = [
    {"order_id": "A1", "item": "Hàng A", "line_amount": 100_000},
    {"order_id": "A1", "item": "Hàng B", "line_amount":  50_000},
    {"order_id": "A1", "item": "Hàng C", "line_amount":  30_000},
]

doanh_thu_tho = sum(r["line_amount"] for r in don_hang)          # đúng: 180.000
doanh_thu_sau_noi = sum(r["line_amount"] for r in chi_tiet)    # sai: 180.000 × 3

print(f"Doanh thu đúng (tổng ở cấp đơn)      : {doanh_thu_tho:>12,.0f} đ")
print(f"Doanh thu nếu cộng sau khi nối 1-N   : {doanh_thu_sau_noi:>12,.0f} đ")
print(f"Hệ số nhân lên                        : {doanh_thu_sau_noi / doanh_thu_tho:.0f} lần")
print()
print("-> Muốn so sánh số đơn và doanh thu giữa các doanh nghiệp khác grain,")
print("   phải tổng hợp dòng hàng lên cấp đơn TRƯỚC khi so sánh.")
```

## Nguyên tắc

> **Dữ liệu ngoại lệ không mặc nhiên là dữ liệu sai**; cần kiểm tra và gắn cỏ trước khi xử lý.

```python
# Minh hoa: quy tac loc don hop le theo nghiem vu (khong huuy, khong tra)
# Don co tra hang (ke ca tra mot phan) -> bo ca don. Don dang giao -> van tinh.

don_hang = [
    # ma don, trang thai, ghi chu, thanh tien
    ("ORD001", "Da giao",    "",                  120_000),
    ("ORD002", "Da huy",     "",                   50_000),
    ("ORD003", "Da giao",    "khach tra lai hang",  80_000),
    ("ORD004", "Dang giao",  "",                   30_000),
    ("ORD005", "Da giao",    "khach mang tra",      70_000),
]

def la_hop_le(trang_thai, ghi_chu):
    """Chỉ tính đơn không huỷ và không trả hàng (kể cả trả một phần)."""
    if trang_thai in ("Da huy", "Da tra hang"):
        return False
    if "tra lai" in ghi_chu.lower():   # tra 1 phan -> bo ca don
        return False
    if "khach mang tra" in ghi_chu.lower():  # ghi chu chua phan loai khac
        return False
    return True

hop_le = [d for d in don_hang if la_hop_le(d[1], d[2])]
doanh_thu = sum(d[3] for d in hop_le)

print(f"Tổng đơn            : {len(don_hang)}")
print(f"Đơn hợp lệ         : {len(hop_le)}")
print(f"Doanh thu hợp lệ    : {doanh_thu:,.0f} đ")
print(f"Tỷ lệ loại          : {(len(don_hang)-len(hop_le))/len(don_hang):.2%}")
print()
for ma, tt, gc, _ in don_hang:
    trang = "HỢP LỆ  " if la_hop_le(tt, gc) else "LOẠI    "
    print(f"  {ma}  {trang}  {tt:<12} {gc}")
```

### Ngoại lệ IQR: tỷ lệ theo từng doanh nghiệp

| Doanh nghiệp | Thành tiền dòng hàng | Số lượng cơ bản | Doanh thu cấp đơn |
|---|---|---|---|

| `dev_009` | — | — | 4,08% |
| `dev_011` | **19,52%** | — (IQR = 0) | **15,67%** |
| `dev_012` | — | — | 4,46% |

**Quyết định xử lý:** giữ nguyên dữ liệu; giữ các đơn hủy/trả trong bảng chất lượng để kiểm tra nhưng **loại khỏi phân tích doanh thu**; giữ phiếu bảo hành 0 đồng; **không tự động loại** ngoại lệ IQR.

> **Lưu ý:** Với `dev_002` và `dev_011`, `qty_base` có **IQR = 0** (phần lớn đơn chỉ có 1 đơn vị cơ bản) → không gắn cờ giá trị lớn hơn 1, tránh báo động giả. Mọi tỷ lệ trên đều là **cờ rà soát**, không phải kết luận dữ liệu sai.

```python
# Minh hoạ: quy tắc IQR — gắn cờ để đối chiếu, KHÔNG xoá
import numpy as np

def vong_ngoai_iqr(values):
    q1, q3 = np.percentile(values, [25, 75])
    iqr = q3 - q1
    return q1, q3, iqr, q1 - 1.5 * iqr, q3 + 1.5 * iqr

# Doanh thu cấp đơn của dev_011: Q1 = 150.000 đồng, Q3 = 3.519.000 đồng
q1, q3 = 150_000, 3_519_000
iqr = q3 - q1
lo, hi = q1 - 1.5 * iqr, q3 + 1.5 * iqr

print(f"Q1              = {q1:>12,.0f} đ")
print(f"Q3              = {q3:>12,.0f} đ")
print(f"IQR = Q3 - Q1   = {iqr:>12,.0f} đ")
print(f"Hàng rào dưới   = {lo:>12,.0f} đ")
print(f"Hàng rào trên   = {hi:>12,.0f} đ")
print()
print("Tỷ lệ đơn ngoài hàng rào trên: 15,67%  ->  đây là CỜ RÀ SOÁT, không phải kết luận lỗi")
```

## Định nghĩa chỉ số và chuẩn hóa

**Chuẩn hoá chỉ số giúp so sánh các mô hình vận hành khác nhau**

## Chỉ số cuối tuần

- **Ngày thường là mốc tham chiếu (100).**
- So sánh số đơn hợp lệ trung bình/ngày cuối tuần với ngày thường trên **cùng cửa sổ**.

$$
I_{\text{cuối tuần}} = 100 \times \frac{\text{Số đơn hợp lệ TB/ngày cuối tuần}}{\text{Số đơn hợp lệ TB/ngày thường}}
$$

| Thuộc tính | Giá trị |
|---|---|
| Phạm vi thời gian | **328 ngày chung** (01/07/2025 – 24/05/2026) |
| Đơn vị | Chỉ số (ngày thường = 100) |

## Biến động Tết 2026

- So với giai đoạn **trước Tết**.
- So sánh số đơn hợp lệ trung bình/ngày của hai cửa sổ có **cùng độ dài**.

$$
A_{\text{ret}} = 100 \times \frac{\text{TB đơn/ngày trong Tết} - \text{TB đơn/ngày trước Tết}}{\text{TB đơn/ngày trước Tết}}
$$

| Thuộc tính | Giá trị |
|---|---|
| Cửa sổ phân tích | **Ba giai đoạn** — Trước Tết · Trong Tết · Sau Tết |
| Độ dài | **17 ngày / cửa sổ** |

## Lưu ý về phạm vi và chuẩn hóa

- Hai chỉ số có **phương pháp chuẩn hoá và phạm vi thời gian riêng**, **không được trộn lẫn tính**.
- Cần đối chiếu: điều kiện đơn hợp lệ, cách tính trung bình/ngày và xử lý khi mẫu số bằng 0.

```python
# Minh hoạ: công thức chỉ số cuối tuần (Ngày thường = 100)
def chi_so_cuoi_tuan(tb_ngay_thuong, tb_ngay_cuoi_tuan):
    """Chỉ số cuối tuần: Ngày thường là mốc tham chiếu = 100."""
    return 100 * tb_ngay_cuoi_tuan / tb_ngay_thuong

# Kiểm chứng công thức trên một cặp số minh hoạ
tb_thuong, tb_cuoi_tuan = 100.0, 155.57
print(f"Ví dụ: TB/ngày ngày thường = {tb_thuong}, TB/ngày cuối tuần = {tb_cuoi_tuan}")
print(f"  -> chỉ số = {chi_so_cuoi_tuan(tb_thuong, tb_cuoi_tuan):.2f}  (cao hơn ngày thường 55,57%)")
print()

# Chỉ số đã tính trên cửa sổ so sánh chung 328 ngày (01/07/2025 – 24/05/2026)
chi_so = [
    ("dev_007", 155.57, "F&B (Nhà hàng / Cà phê)"),
    ("dev_002", 143.50, "F&B (Nhà hàng / Cà phê)"),
    ("dev_011", 136.73, "Phụ kiện & Thiết bị số"),
    ("dev_012", 131.47, "Sửa chữa & Phụ tùng xe"),
    ("dev_009", 108.97, "Chưa xác định (Order-level)"),
    ("dev_006", 105.19, "Nhà thuốc & Dược phẩm"),
]

print(f"{'DN':<10}{'Chỉ số':>9}{'Lift vs ngày thường':>22}   Ngành")
print("-" * 78)
for ma, idx, nganh in chi_so:
    print(f"{ma:<10}{idx:>9.2f}{idx-100:>21.2f}%   {nganh}")
print()
print("-> Cả sáu doanh nghiệp đều > 100, nhưng mức tăng rất khác nhau:")
print("   không có một ngưỡng tăng cuối tuần chung cho mọi doanh nghiệp.")
```

```python
# Minh hoạ: công thức biến động Tết (so sánh hai cửa sổ cùng độ dài 17 ngày)
def bien_dong_tet(tb_truoc_tet, tb_trong_tet):
    """Mức thay đổi % của TB đơn/ngày trong Tết so với trước Tết."""
    return 100 * (tb_trong_tet - tb_truoc_tet) / tb_truoc_tet

# (DN, TB/ngày trước Tết, TB/ngày trong Tết, TB/ngày sau Tết)
tet_2026 = [
    ("dev_009", 35.65, 87.06, 38.29),
    ("dev_011",  8.47, 12.65,  9.59),
    ("dev_012", 14.88, 17.18, 13.88),
    ("dev_002", 107.88, 58.88, 103.76),
    ("dev_007", 56.59, 25.76, 55.00),
    ("dev_006", 79.71, 79.47, 79.00),
]

print(f"{'DN':<10}{'Trước Tết':>12}{'Trong Tết':>12}{'Sau Tết':>11}{'Biến động':>12}")
print("-" * 59)
for ma, truoc, trong, sau in tet_2026:
    print(f"{ma:<10}{truoc:>12,.2f}{trong:>12,.2f}{sau:>11,.2f}{bien_dong_tet(truoc, trong):>11.2f}%")
print()
print("-> Hai chiều: dev_009 tăng mạnh nhất (+144,22%), dev_007 giảm mạnh nhất (-54,47%),")
print("   dev_006 đi ngang tuyệt đối (-0,30%).")
print("-> Nhóm F&B phục hồi ngay sau mùng 10: dev_007 về 55,00 đơn/ngày,")
print("   dev_002 về 103,76 đơn/ngày (gần mức trước Tết).")
```

Kiểm tra nhật ký giao dịch từng ngày xác nhận cả sáu doanh nghiệp đều có dữ liệu liên tục trước và sau kỳ Tết, đồng thời ghi nhận giao dịch đều đặn **17/17 ngày trong Tết** (**0 ngày nghỉ, 0 đơn**) → mức giảm của F&B phản ánh **sự sụt giảm mật độ đơn trên mỗi ngày mở cửa**, không phải do thiếu hụt nguồn dữ liệu.

```python
# Minh hoa: so sanh hai ky Tet cua F&B - mau hinh lap lai qua 2 nam
def bien_dong(tb_truoc, tb_trong):
    return 100 * (tb_trong - tb_truoc) / tb_truoc

# (ky Tet, TB/ngay truoc, TB/ngay trong, TB/ngay sau)
tet = [
    ("Tết 2025 - dev_002", 111.12, 65.35, 110.65),
    ("Tết 2026 - dev_002", 107.88, 58.88, 103.76),
    ("Tết 2025 - dev_009",  33.47, 88.65,  34.35),
    ("Tết 2026 - dev_009",  35.65, 87.06,  38.29),
]

print(f"{'Kỳ Tết':<22}{'Trước':>10}{'Trong':>10}{'Sau':>10}{'Biến động':>12}{'Sau vs trước':>15}")
print("-" * 79)
for nhan, truoc, trong, sau in tet:
    print(f"{nhan:<22}{truoc:>10.2f}{trong:>10.2f}{sau:>10.2f}"
          f"{bien_dong(truoc, trong):>11.2f}%{bien_dong(truoc, sau):>14.2f}%")
print()
print("-> dev_002 giam manh ca 2 ky (-41,19% va -45,42%) roi phuc hoi ve gan")
print("   muc truoc Tet ngay sau mung 10 (-0,42% va -3,82%).")
print("-> dev_009 bung no cung chieu ca 2 ky (+164,85% va +144,22%).")
print("-> Mau hinh lap lai qua 2 ky -> dac tinh kinh doanh, nhieu hon nhieu ngay.")
```
